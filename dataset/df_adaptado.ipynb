# Criando a versão adaptada no nosso DF

Originalmente não existe a coluna de 'CHUVA', e, conforme liberado pelo professor, criamos um novo arquivo csv (que é uma cópia do DF original só que substituindo a coluna numérica de precipitação por uma nova coluna categórica de 'CHUVA': 'SIM' ou 'NAO')

```python
import pandas as pd # importa a biblioteca para manipular os dados
import numpy as np # importa a biblioteca para trabalhar com operações numéricas

caminho = "INMET_NE_PE_A341_CARUARU_01-01-2025_A_31-12-2025.CSV" # indica o arquivo original da estação

# sep separa as colunas por ponto e vírgula
# encoding permite ler os acentos do arquivo
# decimal interpreta a vírgula dos números
# skiprows=8 pula as oito linhas de informações da estação
df_original = pd.read_csv(caminho, sep=";", encoding='iso-8859-1', decimal=",", skiprows=8) # carrega as medições originais em um dataframe

df_original["CHUVA"] = np.where(df_original["PRECIPITAÇÃO TOTAL, HORÁRIO (mm)"] > 0, "SIM", "NAO") # marca sim para precipitação acima de zero e nao nos demais casos

# axis=1 indica que será removida uma coluna
df_original = df_original.drop("PRECIPITAÇÃO TOTAL, HORÁRIO (mm)", axis=1) # remove a precipitação numérica após criar o alvo

# format indica a ordem ano, mês e dia no texto original
df_original['Data'] = pd.to_datetime(df_original['Data'], format='%Y/%m/%d') # converte os textos de data em datas reconhecidas pelo pandas


from pathlib import Path # importa a ferramenta para trabalhar com caminhos de arquivos
pasta_saida = Path('../artefatos/datasets') # define a pasta das cópias geradas
# parents=true cria também as pastas intermediárias
# exist_ok=true aceita uma pasta que já existe
pasta_saida.mkdir(parents=True, exist_ok=True) # cria a pasta de saída sem apagar seu conteúdo
df_original.to_csv( # salva a base adaptada sem sobrescrever a base fornecida
    # index=false evita salvar o índice como uma coluna extra
    # encoding mantém a codificação usada nos arquivos do projeto
    pasta_saida / 'INMET_NE_PE_A341_CARUARU_01-01-2025_A_31-12-2025_ADAPTADO.CSV', index=False, encoding='iso-8859-1' # define o nome e as opções do arquivo de saída
)
print('Cópia gerada:', (pasta_saida / 'INMET_NE_PE_A341_CARUARU_01-01-2025_A_31-12-2025_ADAPTADO.CSV').resolve()) # mostra o caminho completo da cópia gerada
```
