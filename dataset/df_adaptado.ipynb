# Criando a versão adaptada no nosso DF

Originalmente não existe a coluna de 'CHUVA', e, conforme liberado pelo professor, criamos um novo arquivo csv (que é uma cópia do DF original só que substituindo a coluna numérica de precipitação por uma nova coluna categórica de 'CHUVA': 'SIM' ou 'NAO')

```python
import pandas as pd
import numpy as np

caminho = "INMET_NE_PE_A341_CARUARU_01-01-2025_A_31-12-2025.CSV"

df_original = pd.read_csv(caminho, sep=";", encoding='iso-8859-1', decimal=",", skiprows=8)

df_original["CHUVA"] = np.where(df_original["PRECIPITAÇÃO TOTAL, HORÁRIO (mm)"] > 0, "SIM", "NAO")

df_original = df_original.drop("PRECIPITAÇÃO TOTAL, HORÁRIO (mm)", axis=1)

df_original['Data'] = pd.to_datetime(df_original['Data'], format='%Y/%m/%d')

df_original.to_csv('INMET_NE_PE_A341_CARUARU_01-01-2025_A_31-12-2025_ADAPTADO.CSV', index=False, encoding='iso-8859-1')
```
