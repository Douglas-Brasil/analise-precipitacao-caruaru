# Importações

```python
from imblearn.under_sampling import RandomUnderSampler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

caminho = "../dataset/INMET_NE_PE_A341_CARUARU_01-01-2025_A_31-12-2025_PROCESSADO.CSV"
df_processado = pd.read_csv(caminho, sep=",", encoding='iso-8859-1', decimal=".")
```

# Divisão de Dados

Queremos que o nosso modelo tenha muitos registros para ser treinado, então decidimos usar a divisão em 90/10:
- 90% da base pra treino;
- 10% da base pra teste.

O principal fator que nos motivou a fazer isso é que o UnderSimpling será aplicado apenas na base pra treino. Se primeiro fosse o balanceamento e depois da divisão de dados, parte dos registros do teste seriam vazados pro treino da IA e no momento de testar o seu desempenho seria inflado não porque o modelo é bom, mas sim porque já conhecia parte das respostas corretas antes do teste.

Inicialmente pretendíamos usar a divisão 80/20, mas seria apenas 7 mil registros pra treino antes do UnderSimplig. Após aplicar, seria cerca de 900 a 1000 registros pro treino, enquanto o teste teria 20% x 8756 = 1752. Ou seja, aplicando a divisão 80/20 teríamos quase o dobro de dados pro teste do que pro treino. 

```python
X = df_processado.drop(columns=['CHUVA'])
y = df_processado['CHUVA']

X_train, X_test, y_train, y_test = train_test_split(
 X, y, test_size=0.1, random_state=42, stratify=y
)

print(len(X_train), len(X_test))
```

Agora que dividimos os dados podemos usar a técnica de UnderSimplig com segurança:

```python
rus = RandomUnderSampler(random_state=42)
X_train_bal, y_train_bal = rus.fit_resample(X_train, y_train)

y_train_bal.value_counts()
```
