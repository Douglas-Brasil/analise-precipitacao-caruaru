# Importações

```python
from imblearn.under_sampling import RandomUnderSampler # importa a técnica que reduz a quantidade da classe mais comum
from sklearn.model_selection import train_test_split # importa a função que separa treino e teste
import pandas as pd # importa a biblioteca para manipular os dados

caminho = "../dataset/INMET_NE_PE_A341_CARUARU_01-01-2025_A_31-12-2025_PROCESSADO.CSV" # indica onde está a base processada fornecida no projeto
# sep usa vírgula para separar as colunas
# encoding permite ler os acentos do arquivo
# decimal usa ponto nos números decimais
df_processado = pd.read_csv(caminho, sep=",", encoding='iso-8859-1', decimal=".") # carrega a base processada em um dataframe
```

# Divisão de Dados

Queremos que o nosso modelo tenha muitos registros para ser treinado, então decidimos usar a divisão em 80/20:
- 80% da base pra treino;
- 20% da base pra teste.



```python
X = df_processado.drop(columns=['CHUVA']) # separa os atributos usados para classificar a chuva
y = df_processado['CHUVA'] # separa a classe de chuva que o modelo deve aprender

X_train, X_test, y_train, y_test = train_test_split( # divide os atributos e as classes em treino e teste
 # x contém os atributos e y contém as classes
 # test_size=0.2 reserva vinte por cento para teste
 # random_state=42 permite repetir a mesma divisão
 # stratify=y mantém proporções semelhantes das classes
 X, y, test_size=0.2, random_state=42, stratify=y # aplica as opções da divisão dos dados
)

print(len(X_train), len(X_test)) # mostra as quantidades de registros de treino e teste
```

Agora que dividimos os dados podemos usar a técnica de UnderSimplig com segurança:

```python
# random_state=42 permite repetir a mesma seleção
rus = RandomUnderSampler(random_state=42) # cria o balanceador que reduz a classe mais comum
# o teste fica intacto para representar a base original
X_train_bal, y_train_bal = rus.fit_resample(X_train, y_train) # reduz apenas o treino até igualar a quantidade das classes

y_train_bal.value_counts() # confere a quantidade de cada classe no treino balanceado
```
